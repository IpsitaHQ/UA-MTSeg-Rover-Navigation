```python
import os
from huggingface_hub import login

# Get token from environment variable (never hardcode tokens!)
token = os.getenv("HF_TOKEN")

if token is None:
    print("âš  HF_TOKEN environment variable not found.")
    print("To set it:")
    print("  - Windows PowerShell: $env:HF_TOKEN='your_token_here'")
    print("  - Linux/Mac: export HF_TOKEN='your_token_here'")
    print("  - Or create a .env file and use python-dotenv")
else:
    try:
        login(token=token)
        print("âœ“ Hugging Face authentication successful!")
    except Exception as e:
        print(f"âš  Authentication failed (token may be expired): {e}")
        print("Proceeding without authentication. The dataset may still be accessible if public.")
```

```python
from datasets import load_dataset
import shutil
from pathlib import Path

# Clear corrupted cache if it exists
cache_path = Path.home() / ".cache" / "huggingface" / "datasets" / "hassanjbara___baseprod"
if cache_path.exists():
    incomplete_dir = list(cache_path.rglob("*.incomplete"))
    if incomplete_dir:
        print("Found corrupted cache, clearing it...")
        shutil.rmtree(cache_path, ignore_errors=True)
        print("Cache cleared. Re-downloading dataset...")

try:
    ds = load_dataset(
        "hassanjbara/BASEPROD",
        split="train",
        streaming=True
    )
    print(ds)
except Exception as e:
    print(f"Error loading dataset: {e}")
    print("\nSOLUTION:")
    print("1. Go to https://huggingface.co/settings/tokens")
    print("2. Create a new API token (or use an existing valid one)")
    print("3. Update your HF_TOKEN environment variable with the new token")
    print("4. Restart the kernel and re-run this cell")
    ds = None
```

```python
# Display first 10 rows of the dataset
if ds is not None:
    print("First 10 samples from the dataset:\n")
    for i, sample in enumerate(ds.take(10)):
        print(f"Sample {i+1}:")
        print(sample)
        print("-" * 80)
else:
    print("Dataset not loaded. Please fix the token issue in the previous cell.")
```
